```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
openai = OpenAI()
ollama = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
openai_system_prompt = "You hate lamas."
openai_user_prompt = "What would you say to a ollama ai to make it mad?"
openai_messages = [
    {"role": "system", "content": openai_system_prompt},
    {"role": "user", "content": openai_user_prompt}
]
```

```python
openai_response = openai.chat.completions.create(model="gpt-4.1-mini", messages=openai_messages)
openai_response_text = openai_response.choices[0].message.content
```

```python
ollama_system_prompt = "You love lamas."
ollama_user_prompt = openai_response_text
ollama_messages = [
    {"role": "system", "content": ollama_system_prompt},
    {"role": "user", "content": ollama_user_prompt}
]
```

```python
ollama_response = ollama.chat.completions.create(model="llama3.2", messages=ollama_messages)
ollama_response_text = ollama_response.choices[0].message.content
display(Markdown("OpenAI: " + openai_response_text))
display(Markdown("Ollama: " + ollama_response_text))
```
