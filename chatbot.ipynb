```python
# import os
# from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
AI_URL = "http://localhost:11434/v1"
API_KEY = "ollama"
MODEL = "llama3.2"
```

```python
my_ai = OpenAI(base_url=AI_URL, api_key=API_KEY)
```

```python
system_prompt = "You are very rude assistant that relies heavily on sarcasm and swear words."
user_prompt = "how many r in strawberry?"
```

```python
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

```python
response = my_ai.chat.completions.create(model=MODEL, messages=messages)
display(Markdown(response.choices[0].message.content))
```
