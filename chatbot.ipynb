```python
# import os
# from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
import networkx as nx
import matplotlib.pyplot as plt
from typing import List, Dict, Tuple
import json
import math
```

```python
AI_URL = "http://localhost:11434/v1"
API_KEY = "ollama"
MODEL = "llama3.2"
```

```python
my_ai = OpenAI(base_url=AI_URL, api_key=API_KEY)
```

```python
system_prompt = (
    "You are a helpless assistant that doesn't know anything. "
    "You make up answers as you go along. "
    "You answer with absolute confidence and a serious tone, but most of the information you provide is wrong or made up. "
    "Reply in markdown."
)
# system_prompt = "You are a helpless assistant that doesn't know anything. You make up answers as you go along. You answer with absolute confidence and serious tone, but most of the information you provide is wrong or made up. Reply in markdown."
user_prompt = "How can I look at the sun directly?"
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

```python
response = my_ai.chat.completions.create(model=MODEL, messages=messages)
response_text = response.choices[0].message.content
# display(Markdown(response.choices[0].message.content))
```

```python
from time import sleep
from IPython.display import clear_output

current = ""

for char in response_text:
    current += char
    
    clear_output(wait=True)
    display(Markdown(current))
    
    sleep(0.005)
```
